```python
import pandas as pd
import joblib
import numpy as np

model = joblib.load("house_price_xgb.pkl")
test = pd.read_csv("houses_test.csv")

test["Room_Density"] = (
    test["area"] /
    test["num_rooms"].replace(0, np.nan)
)

test["is_top_floor"] = (
    test["floor"] == test["max_floor"]
).astype(int)

test = test.drop(
    ["region", "url", "num_rooms", "max_floor", "num_bathrooms"],
    axis=1,
    errors="ignore"
)

pred_log = model.predict(test)
pred_price = np.expm1(pred_log)

submission = pd.DataFrame({
    "price": pred_price
})

submission.to_csv("predictions.csv", index=False)

print(submission.head())
```
